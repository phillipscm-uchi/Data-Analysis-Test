```python
import textkit
from textkit import slugify, word_count, truncate, is_palindrome

print(textkit.__version__)                       # 0.1.0
print(slugify("Héllo, World!"))                  # hello-world
print(word_count("a  b\nc"))                     # 3
print(truncate("abcdefghij", 3))                 # abc...
print(is_palindrome("A man, a plan, a canal: Panama"))  # True
```
